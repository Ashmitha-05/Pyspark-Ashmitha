```python
from pyspark.sql import SparkSession

data = [
    (1, " Ash ", "ash@gmail.com", "Chennai"),
    (2, "JOHN", "john@gmail.com", "Bangalore"),
    (3, "emma", "emma@gmail.com", "Mumbai"),
    (4, "DaVid", "david@gmail.com", "Delhi")
]

df = spark.createDataFrame(
    data,
    ["ID","Name","Email","City"]
)

df.show()
```

```python
from pyspark.sql.functions import *

df.select(
    initcap(trim(lower(col("Name")))).alias("Name")
)
```

```python
df.select(split(col("email"),"@")[0])
```

```python
df.select(split(col("empid"),"-")[2])
```

```python
df.select(regexp_replace(col("phone"),"-",""))
```

```python
df.select(concat_ws(" ",col("first_name"),col("last_name")))
```

```python
df.select(
    regexp_replace(
        col("Department"),
        "[a-z]",
        ""
    ).alias("DeptCode")
)
```

```python
df.withColumn(
    "Name",
    initcap(col("Name"))
)
```

```python
df.withColumn(
    "Name",
    trim(col("Name"))
)
```

```python
df.withColumn(
    "Salary",
    round(col("Salary"),2)
)
```

```python
df.withColumn(
    "Salary",
    abs(col("Salary"))
)
```

```python
df.withColumn(
    "Highest",
    greatest(col("Salary"),
             col("Bonus"))
)
```

```python
df.withColumn(
    "Year",
    year(col("JoinDate"))
)
```

```python
df.withColumn(
    "DaysWorked",
    datediff(current_date(),
             col("JoinDate"))
)
```

```python
df.withColumn(
    "Name",
    trim(col("Name"))
).withColumn(
    "Email",
    lower(col("Email"))
).withColumn(
    "JoinYear",
    year(col("JoinDate"))
)
```

```python
df.withColumn(
    "Phone",
    regexp_replace(col("Phone"),"-","")
).withColumn(
    "Name",
    upper(col("Name"))
).withColumn(
    "Salary",
    round(col("Salary"),0)
)
```
