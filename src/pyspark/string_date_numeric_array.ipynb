```python
from pyspark.sql import SparkSession

data = [
    (1, " Ash ", "ash@gmail.com", "Chennai"),
    (2, "JOHN", "john@gmail.com", "Bangalore"),
    (3, "emma", "emma@gmail.com", "Mumbai"),
    (4, "DaVid", "david@gmail.com", "Delhi")
]

df = spark.createDataFrame(
    data,
    ["ID","Name","Email","City"]
)

df.show()
```

```python
from pyspark.sql.functions import *

df.select(
    initcap(trim(lower(col("Name")))).alias("Name")
)
```

```python
df.select(split(col("email"),"@")[0])
```

```python
df.select(split(col("empid"),"-")[2])
```

```python
df.select(regexp_replace(col("phone"),"-",""))
```

```python
df.select(concat_ws(" ",col("first_name"),col("last_name")))
```

```python
df.select(
    regexp_replace(
        col("Department"),
        "[a-z]",
        ""
    ).alias("DeptCode")
)
```

```python
df.withColumn(
    "Name",
    initcap(col("Name"))
)
```

```python
df.withColumn(
    "Name",
    trim(col("Name"))
)
```

```python
df.withColumn(
    "Salary",
    round(col("Salary"),2)
)
```

```python
df.withColumn(
    "Salary",
    abs(col("Salary"))
)
```

```python
df.withColumn(
    "Highest",
    greatest(col("Salary"),
             col("Bonus"))
)
```

```python
df.withColumn(
    "Year",
    year(col("JoinDate"))
)
```

```python
df.withColumn(
    "DaysWorked",
    datediff(current_date(),
             col("JoinDate"))
)
```

```python
df.withColumn(
    "Name",
    trim(col("Name"))
).withColumn(
    "Email",
    lower(col("Email"))
).withColumn(
    "JoinYear",
    year(col("JoinDate"))
)
```

```python
df.withColumn(
    "Phone",
    regexp_replace(col("Phone"),"-","")
).withColumn(
    "Name",
    upper(col("Name"))
).withColumn(
    "Salary",
    round(col("Salary"),0)
)
```

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()

data = [
    (101, "Alice", "Python", "SQL", "Spark", "Java"),
    (102, "Bob", "Java", "Kafka", "Scala", "Python"),
    (103, "Charlie", "SQL", "Power BI", "Excel", "SQL"),
    (104, "David", "C++", "Python", "AWS", "Docker"),
    (105, "Emma", "Spark", "Databricks", "Python", "Azure")
]

columns = [
    "EmpID",
    "Name",
    "Skill1",
    "Skill2",
    "Skill3",
    "Skill4"
]

df = spark.createDataFrame(data, columns)

df.show(truncate=False)
```

```python
from pyspark.sql.functions import *

dff=df.select("name",array("skill1","skill2","skill3","skill4").alias("skills"))
```

```python

dff.select(col('name'),col('skills')[0].alias("first_skill")).show()
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
dff.select(element_at(col('skills'),-1).alias("last_skill")).show()
```

```python
dff.select(col('skills')[1].alias("secondary_skills")).show()
```

```python
dff.select(size(col('skills').alias("total_count"))).show()
```

```python
dff.select("name',array_contains("skills","python").alias("basic_skills")).show()
```

```python
dff.select(reverse(col('skills')).alias("reverse_skills")).show()
```
