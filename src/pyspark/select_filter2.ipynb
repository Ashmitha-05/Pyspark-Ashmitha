```python
from pyspark.sql.functions import *
df=spark.read.parquet("/Volumes/pyspark/default/volume/large_employee_Parquet/")
df=df.filter((col("salary")>5000) & (col("salary")<10000)) \
    .filter(col("city")=="chennai") \
    .select("name")
df.show()
```

```python
from pyspark.sql.functions import *
df=spark.read.parquet("/Volumes/pyspark/default/volume/large_employee_Parquet/")
df.filter(col("city").isin("chennai","bangalore")) \
    .filter(~col("age").isin(60)) \
    .filter(col("name")=='emma') \
    .where("salary isnotnull")
```
