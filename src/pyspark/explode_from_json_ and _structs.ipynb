```python
from pyspark.sql import Row

data = [
    (
        "Alice",
        [
            Row(ProjectName="HRMS", Duration=12),
            Row(ProjectName="Payroll", Duration=8)
        ]
    ),
    (
        "Bob",
        [
            Row(ProjectName="Inventory", Duration=10)
        ]
    )
]

df = spark.createDataFrame(data, ["Name", "Projects"])

df.show()
```

```python
from pyspark.sql.functions import *

df.select(
    "Name",
    explode("Projects").alias("Project")
).show(truncate=False)
```

```python
df.select(
    "Name",
    inline("Projects")
).show()
```
