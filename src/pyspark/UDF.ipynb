```python
from pyspark.sql.functions import *
from pyspark.sql.types import *

data = [
    ("Alice",45000),
    ("Bob",90000),
    ("Charlie",60000)
]

df = spark.createDataFrame(data,["Name","Salary"])


def grade(salary):

    if salary > 70000:
        return "Gold"

    elif salary > 50000:
        return "Silver"

    else:
        return "Bronze"


grade_udf = udf(grade,StringType())

df.withColumn(
    "Grade",
    grade_udf(col("Salary"))
).show()
```

```python
def full_name(first, last):
    return first + " " + last

full_name_udf = udf(full_name, StringType())

df.withColumn(
    "FullName",
    full_name_udf(
        col("FirstName"),
        col("LastName")
    )
)
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
def adhaar_no(no):
    return "xxxxx"+no[-4:]

adhaar_udf=udf(adhaar_no,StringType())

df.withColumn(
    "name",adhaar_no(col('adhaar'))
)

```
