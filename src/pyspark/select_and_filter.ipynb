```python
from pyspark.sql.types import *
from pyspark.sql.functions import *

df=spark.read.csv("/Volumes/pyspark/default/volume/large_employee_500k.csv",header=True,inferSchema=True)

df.select("Name")
df.select("Name", "Salary")
df.select(
    "Name",
    "Salary",
    "Dept"
)
df.select(
    col("Salary").alias("EmployeeSalary")
)
df.select("*")
df.selectExpr(
    "Salary * 12 as AnnualSalary"
)

df.filter(
    col("Salary") > 50000
)
df.filter(
    col("Dept") == "IT"
)
df.filter(
    (col("Salary") > 50000) &
    (col("Age") > 25)
)
df.filter(
    (col("Dept") == "HR") |
    (col("Dept") == "Finance")
)
df.filter(
    col("Dept").isin("HR", "IT")
)
df.filter(
    ~col("Dept").isin("HR", "IT")
)
df.filter(
    col("Salary").between(50000, 100000)
)
df.filter(
    col("Name").startswith("A")
)
df.filter(
    col("Name").endswith("a")
)
```

```python
df.withColumn(
    "Bonus",
    col("Salary") * 0.10
)
df.withColumn(
    "AnnualSalary",
    col("Salary") * 12
)
df.withColumn(
    "Salary",
    col("Salary") + 5000
)
df.withColumn(
    "Tax",
    col("Salary") * 0.05
)
df.withColumn(
    "Dept",
    upper(col("Dept"))
)
df.withColumn(
    "NameLength",
    length(col("Name"))
)

df.withColumn(
    "Grade",
    when(col("Salary") >= 100000, "High")
    .when(col("Salary") >= 50000, "Medium")
    .otherwise("Low")
)
df.withColumn(
    "Adult",
    when(col("Age") >= 18, "Yes")
    .otherwise("No")
)
```

```python
df.groupBy("department").agg(
    sum("salary")
)
df.groupBy("department").agg(
    avg("salary")
)
df.groupBy("department").agg(
    max("salary")
)
df.groupBy("department").agg(
    min("salary")
)
df.groupBy("department").agg(
    count("*")
)
df.groupBy("department").agg(
    sum("salary"),
    avg("salary"),
    max("salary"),
    min("salary"),
    count("*")
)
df.groupBy("department").agg(
    sum("salary").alias("TotalSalary"),
    avg("salary").alias("AverageSalary")
)
df.groupBy("department").agg(
    countDistinct("name")
)
df.groupBy("department").agg(
    first("name")
)
df.groupBy("department").agg(
    last("name")
)
df.groupBy("department").agg(
    collect_list("name")
)
df.groupBy("department").agg(
    collect_set("name")
)
df.agg(
    sum("salary"),
    avg("salary")
)
```

```python


department_data = [
    ("Finance", "Ravi", "Chennai"),
    ("HR", "Kiran", "Mumbai"),
    ("IT", "Anita", "Bengaluru"),
    ("Marketing", "Rahul", "Delhi"),
    ("Sales", "Sneha", "Hyderabad")
]

department_df = spark.createDataFrame(
    department_data,
    ["department", "manager", "head_office"]
)

df.join(
    department_df,
    "department",
    "inner"
)



df.join(
    department_df,
    "department",
    "left"
)



df.join(
    department_df,
    "department",
    "right"
)




df.join(
    department_df,
    "department",
    "full"
)



df.join(
    department_df,
    "department",
    "outer"
)




df.join(
    department_df,
    "department",
    "left_semi"
)




df.join(
    department_df,
    "department",
    "left_anti"
)



df.join(
    department_df,
    how="cross"
)




df.join(
    department_df,
    df.department == department_df.department,
    "inner"
)




df.join(
    department_df,
    ["department", "city"],
    "inner"
)




df.join(
    department_df,
    "department"
).select(
    "name",
    "department",
    "manager",
    "salary"
)




emp = df.alias("emp")
dept = department_df.alias("dept")

emp.join(
    dept,
    col("emp.department") == col("dept.department"),
    "inner"
).select(
    col("emp.id"),
    col("emp.name"),
    col("emp.department"),
    col("emp.salary"),
    col("dept.manager"),
    col("dept.head_office")
)
```
