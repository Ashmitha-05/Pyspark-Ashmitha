```python
from pyspark.sql.types import *
from pyspark.sql.functions import *
df=spark.read.csv("/Volumes/pyspark/default/volume/large_employee_500k.csv",header=True,inferSchema=True)
df.select("name")

df.select("name","Salary")

df.select("*")

df.select(col("name"))

df.select(df.Name)

df.select(col("name").alias("Employee_Name"))

df.select((col("salary")*1.1).alias("NewSalary"))


cols=["Name","Salary"]
df.select(*cols)
```
