```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()

data = [
(101,"Ash","IT","Chennai",50000,5000),
(102,"Rahul","IT","Bangalore",65000,7000),
(103,"John","HR","Chennai",45000,None),
(104,"Priya","HR","Bangalore",55000,4000),
(105,"David","Finance","Mumbai",70000,9000),
(106,"Sam","Finance","Mumbai",65000,None),
(107,"Emma","IT","Chennai",80000,10000),
(108,"Sophia","HR","Delhi",60000,6000),
(109,"Mike","Finance","Delhi",75000,8000),
(110,"Chris","IT","Bangalore",90000,None),
(111,"Tom","Sales","Chennai",40000,3000),
(112,"Jerry","Sales","Delhi",42000,3500)
]

columns=["EmpID","Name","Dept","City","Salary","Bonus"]

df=spark.createDataFrame(data,columns)
```

```python
from pyspark.sql.functions import *
df.groupBy(col("Dept")).agg(count(col("empid")))
```

```python
df.groupBy(col("Dept")).agg(sum("salary"))
```

```python
df.groupBy(col("Dept")).agg(mean("salary"))
```

```python
df.groupBy(col("dept")).agg(min("salary"),max("salary"))
```

```python
df.groupBy(col("dept")).agg(count("Bonus"))
```

```python
df.groupBy(col("dept")).agg(count("*"))
```

```python
df.groupBy(col("dept")).agg(collect_list("name"),collect_set("name"))
```

```python
df.groupBy(col("dept")).agg(first("name")).orderBy("empid")
```

```python
df.groupBy(col("dept")).agg(last("name")).orderBy(col("empid").desc())
```

```python
df.groupby(col("dept")).agg(sum('salary')).filter(col('salary')>150000)
```

```python
df.groupBy(col("city")).agg(count('empid')).alias("emp_count").filter(col('emp_count')>2)
```

```python
df.groupBy(col("city"),col("dept")).agg(sum('salary')).alias("totoal_sal").filter(col('total_sal')>100000)
```

```python
df.groupby(col("dept")).agg(sum('salary').alias("total_sal")).orderBy(col("total_sal").desc())
```
