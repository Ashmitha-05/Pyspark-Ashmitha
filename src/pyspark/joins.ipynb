```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()

emp_data = [
(101,"Ash",1,50000),
(102,"Rahul",2,65000),
(103,"John",1,45000),
(104,"Priya",3,70000),
(105,"Emma",4,80000),
(106,"Chris",None,55000)
]

emp_columns = ["EmpID","Name","DeptID","Salary"]

emp = spark.createDataFrame(emp_data, emp_columns)
```

```python
dept_data = [
(1,"IT"),
(2,"HR"),
(3,"Finance"),
(5,"Sales")
]

dept_columns = ["DeptID","Department"]

dept = spark.createDataFrame(dept_data, dept_columns)
```

```python
emp.join(dept,"DeptID","inner")
```

```python
dd = emp.join(
    dept,
    "DeptID",
    "inner"
).select(
    "Name",
    "Department"
)

dd.show()
```

```python
dd = emp.join(
    dept,
    "DeptID",
    "inner"
).filter(
    col("Department") == "IT"
)

dd.show()
```

```python
emp.join(dept,"DeptID","left")
```

```python
emp.join(dept,"DeptID","left").select("name","department")
```

```python
from pyspark.sql.functions import *
emp.join(dept,'DeptID',"left").filter(col('department').isnull())
```

```python
emp.join(dept,"DeptID","left").groupBy(col("department")).agg(count('empid')).show()
```
